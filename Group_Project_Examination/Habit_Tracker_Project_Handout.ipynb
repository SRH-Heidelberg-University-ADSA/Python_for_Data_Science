### 1.1 The shared data format

All five parts use the **same dictionary**. It is called `data` everywhere.

```python
data = {
    "habits": {
        "1": {"name": "Read 20 minutes", "created": "2026-10-05"}
    },
    "log": {
        "1": ["2026-10-05", "2026-10-06"]   # dates when habit "1" was done
    }
}
```

- Habit ids are **text** ("1", "2", ...), not numbers.
- Dates are text in the format **YYYY-MM-DD**: 4-digit year, 2-digit month, 2-digit day, with dashes, exactly 10 characters. Example: `2026-01-05`. These are **not valid**: `2026-1-5`, `20261005`, `2026-W41-1`, `05.01.2026`.
- `data["log"]` may have no entry for a habit that was never done.

<a id="parse-date"></a>
### 1.2 Checking dates: `parse_date`

**Roles B, C, D, and E** check dates with the same function. You may not import a teammate's module, so **copy this function into your own file** and do not change it. Add `import datetime` at the top of the file.

```python
def parse_date(text):
    """Return a datetime.date for 'YYYY-MM-DD', or None if invalid."""
    if not isinstance(text, str) or len(text) != 10:
        return None
    try:
        return datetime.datetime.strptime(text, "%Y-%m-%d").date()
    except ValueError:
        return None
```

> **Important note:** do not use `fromisoformat()` to check dates. Newer Python versions also accept other formats such as `20261005`.

| Function | What it does |
|---|---|
| `load_data(path) -> dict` | Reads a JSON file and returns `data`. If the file is **missing or broken**, return `{"habits": {}, "log": {}}`. Never crash. |
| `save_data(data, path) -> bool` | Writes `data` to a JSON file. Returns `True` if it worked, `False` if not (for example, the folder does not exist). |

### Role C: Analytics (`analytics.py`)

- **Your job:** turn the log into numbers.
- **Main topics:** loops, conditions, dates, sets, pandas/NumPy.

| Function | What it does |
|---|---|
| `current_streak(data, habit_id) -> int` | Number of days in a row, ending **today**. If today is not done yet, count the run that ends **yesterday**. If neither today nor yesterday is done, return 0. |
| `longest_streak(data, habit_id) -> int` | The longest run of days in a row, ever. The log is **not sorted**, so you must sort it. |
| `completion_rate(data, habit_id, days) -> float` | Calculates how often the habit was completed over the past `days` calendar days, including today. Returns the result as a decimal between 0.0 and 1.0. |
| `weekly_summary(data) -> dict` | For **every** habit: how many times it was done in the last 7 days (today and the 6 days before). Example: `{"1": 5, "2": 0}`. |

**Special rules**

- A habit id is **unknown** when it is not a key of `data["habits"]`, even if `data["log"]` has entries for it. For an unknown id, return `0` from `current_streak` and `longest_streak`, and `0.0` from `completion_rate`. If `days` is 0 or less, return `0.0`.
- `weekly_summary` has exactly one key for every habit in `data["habits"]`. Log entries of ids that are not habits are ignored.
- Ignore log entries that are not valid dates (do not crash). Check each log entry with your copy of [`parse_date`](#parse-date); if it returns `None`, ignore the entry. Also ignore dates in the future, and count every date only once (duplicates in the log are ignored).
- `completion_rate` always divides by `days`, even if the habit is younger than `days` (was created fewer than days). Do not round the result (`10 / 30` stays `0.3333...`).
- Use `datetime.timedelta` for the day calculations. Do not use `fromisoformat()`.
- **Common problem:** if you use NumPy/pandas, wrap results: `int(...)`, `float(...)`.

**Done, tests, tasks**

- **Done when:** all edge cases work (empty log, gaps, streak ended yesterday, unsorted dates).
- **Your tests:**
  - Unit tests: at least 5, in `test_analytics.py`.
  - Integration test (`test_integration_C.py`): build data with `habits.add_habit` and `habits.mark_done` (mark the dates in an unsorted order, and check that the stored list in `data["log"]` really is unsorted; if `mark_done` sorts it, write an unsorted list into `data["log"]` yourself), then check the streaks and the weekly summary, check that your functions do not change `data` (compare with `copy.deepcopy`), and check that the numbers are the same after `storage.save_data` and `storage.load_data`.

### 4.1 Git without clashes

Five people work in one repository. Most problems can be avoided with a few simple rules.

**One-time setup (everybody, once):**

```
git config --global pull.rebase false
```

Without this, new versions of Git can stop with the error "Need to specify how to reconcile divergent branches".

### 4.2 Integration

Integration means joining the five modules into one working app. It is designed to be simple: **swap the stub for the real file, then test.** You are not merging five different programs.

**Why it works**

1. **One folder, one owner per file.** Nobody edits another student's file, so there are no code conflicts. The shared files have one section per student (`README.md`, `progress.md`) or one signature block per student (`contract.md`). **Always run `git pull` before `git push`.**
2. **`main.py` does not change when a stub is replaced.** Role E builds the menu against the stubs. A stub and the real function have the same name, the same parameters, and the same return type, so the real file simply takes the place of the stub.
3. **Roles A to D never import each other inside their modules.** Only Role E's `main.py` uses all four modules in the app, and the five integration test files (one per student) use them to test the fit. So each module can be built, tested, and checked alone.
4. **One `data` dictionary travels through the whole app.** Role E loads it through A, passes it to B to change it, to C to get numbers, and to D to build a report, and saves it through A at the end. Nobody keeps a private copy.

**The six steps (first part of Week 7).** The work is shared: **nobody does the integration alone**.

| Step | Who | What |
|---|---|---|
| 1 | Everyone | Run `git pull`. All real modules are now in one folder and the old stubs are replaced. |
| 2 | Role C runs the check. The owners fix. | Role C runs `python check_contract.py all full` and `python check_contract.py main full` and posts both outputs. The owner of each failing module fixes their **own** file and runs the checker again. |
| 3 | Everyone | Run your own integration test (`python test_integration_X.py`). If it fails because of **your** code, fix it. If it fails because of a teammate's code, send a [bug report](#bug-reports). |
| 4 | Everyone | Go through the deadlines in [Part 6](#part-6). Each student has to check the deliverables for each milestone. |
| 5 | The whole group | Fix problems that only appear when the parts work together (see [typical problems](#typical-problems)). Each problem goes to the **owner** of the file. |
| 6 | The whole group | **Freeze** the code. Everybody checks their part of the list below, then you rehearse the demo and submit. |

**Freeze checklist (one part per student).** The whole group completes it.

- **Role A:** no `# STUB` line is left in a file that has real code, and `requirements.txt` matches the imports.
- **Role B:** the group section of `progress.md` is complete.
- **Role C:**  run `python check_contract.py all full` and `python check_contract.py main full` one last time; both pass.

## 5. Challenge: design patterns

Your **group** implements **two** of these patterns: Singleton, Factory Method, Strategy, Observer (10 points: 4 per implemented pattern and 1 per rationale, see the Project Examination PDF, Section 5.4). To keep the roles independent, follow these rules:

- A pattern counts as implemented when (1) it is a **class** in one student's own file, (2) a **contract function** uses it, and (3) `python check_contract.py all full` still passes.
- `all full` checks `storage.py`, `habits.py`, `analytics.py`, `reports.py` and `utils.py`, but **not** `main.py`. Put a pattern in one of these five files, so that the checker can prove that it still works.
- The group decides which pattern fits which file, and **all five students must approve** the decision (write it in the pattern table of `contract.md`). The **owner of the file writes the code** (nobody edits another student's file) and explains it in the Q&A. The two patterns may be in the same file or in two files.
- A pattern must **not change** the function names, parameters, or return types of the contract. The contract functions stay normal functions.
- Roles A to D do not import each other's modules, so each pattern must work inside its own file.
- Add the rationale for each pattern in `design_patterns.md`. The owner of the file writes that pattern's section.

For each pattern, the rationale answers three questions:

1. Which pattern, and in which file and function did you use it?
2. What problem does it solve in this app?
3. What would be harder without it?

Ideas that fit this app:

| Pattern | What it means in one line | Idea for this app |
|---|---|---|
| **Strategy** | Swap one algorithm for another behind the same call. | `reports.py`: a text chart and an image chart as two interchangeable "renderers" used by `build_report` and `save_chart_image`. |
| **Factory Method** | One function decides which object to create. | `reports.py`: `make_chart(kind)` returns the right chart object, and `build_report` or `save_chart_image` uses it. |
| **Observer** | Objects subscribe and are told when something happens. | `habits.py`: a small event class. `mark_done` tells its listeners (for example a counter of marks) after every successful mark. |
| **Singleton** | Exactly one shared instance of a class. | `analytics.py`: one `Clock` object that returns today's date, and every analytics function asks it. (It must give the real date every time it is asked, because the checker works with dates relative to today.) Do **not** keep the file path in a Singleton: `load_data(path)` and `save_data(data, path)` get a different path in every call. |

## Appendix

### A. Glossary

| Word | Meaning |
|---|---|
| **Contract** | The fixed list of function names, parameters, return types, and the `data` format that all five parts share. |
| **Module** | One Python file with one role's functions (for example `habits.py`). |
| **Stub** | A fake module with the right function names and fixed return values. It starts with `# STUB`. |
| **Quick check** | `python check_contract.py <module>`: do the functions exist and return the right type? |
| **Full check** | `python check_contract.py <module> full`: are the results correct? |
| **Integration** | Joining the five modules into one working app. |
| **Integration test** | A test (`test_integration_X.py`) that checks that your module works together with the others. |
| **Freeze** | The moment when the code stops changing before submission. |
| **Peer review** | Reading and commenting on a teammate's work, without editing it. |
| **Design pattern** | A known, reusable way to structure code to solve a common problem. |

### B. FAQ

| Question | Answer |
|---|---|
| What if my teammate does not deliver a real module? | The group keeps that student's stub, so the app still starts. The group mark goes down, but the individual marks of the other four students are not affected. See [the contract rules](#part-1). |
| Can I use AI tools? | Yes, to **learn new topics** and to help you **fix logical or syntax errors**. You must write in your note which tools you used and how, and you must be able to explain every line of your own code in the oral check. If you cannot, it may affect your Q&A grade, and in serious cases it counts as an academic integrity violation. |
| Can I change the contract after Week 3? | Only if all five students agree, only to **add** something, and it must be written in the change log in `contract.md`. Never rename or remove a function. |
| I need `matplotlib`, `pandas`, or `numpy`. | Only Role D may use `matplotlib`, and only Role C may use `pandas` and `numpy`. Tell Role A so it goes into `requirements.txt`. |
| My integration test fails because of a teammate's code. | Send a [bug report](#bug-reports): function, input, expected result, real result. You lose no marks for that. |
| `git push` is rejected. | This is normal. Run `git pull`, then `git push` again. |
| `ModuleNotFoundError: No module named 'storage'` | The file is missing, or you are in the wrong folder. |
| `import csv` gives a strange error. | Look for a file called `csv.py` in your folder and rename it. |

<details>
<summary><b>Example for Role B (<code>test_integration_B.py</code>): click to open</b></summary>

```python
# test_integration_B.py: Role B's module together with analytics, reports, storage
import datetime
import habits, analytics, reports, storage

TODAY = datetime.date.today().isoformat()

def test_marked_habit_has_a_streak():
    data = {"habits": {}, "log": {}}
    habit_id = habits.add_habit(data, "Read")
    habits.mark_done(data, habit_id, TODAY)
    assert analytics.current_streak(data, habit_id) == 1

def test_removed_habit_is_gone_everywhere():
    data = {"habits": {}, "log": {}}
    keep_id = habits.add_habit(data, "Read")
    gone_id = habits.add_habit(data, "Swim")
    habits.mark_done(data, keep_id, TODAY)
    habits.mark_done(data, gone_id, TODAY)
    habits.remove_habit(data, gone_id)
    assert gone_id not in analytics.weekly_summary(data)
    assert keep_id in analytics.weekly_summary(data)
    report = reports.build_report(data)
    assert "Swim" not in report
    assert "Read" in report

def test_ids_stay_text_after_saving():
    data = {"habits": {}, "log": {}}
    habit_id = habits.add_habit(data, "Read")
    assert isinstance(habit_id, str)
    storage.save_data(data, "_test_int_B.json")
    loaded = storage.load_data("_test_int_B.json")
    assert habit_id in loaded["habits"]

test_marked_habit_has_a_streak()
test_removed_habit_is_gone_everywhere()
test_ids_stay_text_after_saving()
print("Integration test B passed")
```

<details>
<summary><b>Example for Role E (<code>test_integration_E.py</code>): click to open</b></summary>

The app runs as a separate program and receives a fixed list of answers. In this example, the empty answer after the habit id means "today" for ask_date, and the "n" answers the question "Save a chart image?" that appears after the report.

```python
# test_integration_E.py: a scripted user journey through main.py
import datetime
import os
import subprocess
import sys
import storage

TEST_FILE = "_test_int_E.json"
TODAY = datetime.date.today().isoformat()

def run_app(answers):
    if os.path.exists(TEST_FILE):
        os.remove(TEST_FILE)
    return subprocess.run(
        [sys.executable, "main.py", TEST_FILE],
        input="\n".join(answers) + "\n",
        capture_output=True, text=True, timeout=30,
    )

def test_add_mark_report_quit():
    result = run_app(["1", "Read", "3", "1", "", "6", "n", "0"])
    assert result.returncode == 0
    assert "Traceback" not in result.stderr
    data = storage.load_data(TEST_FILE)
    assert data["habits"]["1"]["name"] == "Read"
    assert TODAY in data["log"]["1"]

def test_wrong_input_does_not_crash():
    result = run_app(["abc", "99", "", "3", "42", "", "0"])
    assert result.returncode == 0
    assert "Traceback" not in result.stderr
    assert os.path.exists(TEST_FILE)

test_add_mark_report_quit()
test_wrong_input_does_not_crash()
print("Integration test E passed")
```

(The answers follow the fixed order of questions in the Role E rules. The checker uses the same order.)
